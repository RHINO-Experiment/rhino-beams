# Example of Upscaling Beams

```python
import numpy as np
import matplotlib.pyplot as plt
import healpy as hp
```

```python
from beam_scripts.beam_upscale_cst import upscale_cst_beams_and_save_npz
```

```python
upscale_cst_beams_and_save_npz(npz_filepath='npz_beams/DipoleWet.npz',
                               save_dir='test_dir',
                               cst_i=False,
                               save_hp=True,
                               beam_label='DipoleWet',
                               n_side=None)
```

```python
test_beam = hp.fitsfunc.read_map('test_dir/DipoleWet59.0.fits')
hp.mollview(test_beam)
plt.show()
hp.mollview(10*np.log10(test_beam))
plt.show()
```
